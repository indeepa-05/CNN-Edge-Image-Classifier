**Observing Validation Metrics it is clear that the model is overfitting. The model is performing well on the training data but poorly on the validation data. This indicates that the model has learned to memorize the training data rather than generalizing to unseen data.**

```python
#Implementing LR scheduler to reduce learning rate on plateau
tf.keras.backend.clear_session()
tf.keras.utils.set_random_seed(42)
```

```python
ModelA = build_model(verbose=False)

ModelA.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss=tf.keras.losses.SparseCategoricalCrossentropy(),
    metrics=["accuracy"]
)

ModelA.summary()
```

```python
# Reduce learning rate when validation loss stops improving
lr_scheduler = tf.keras.callbacks.ReduceLROnPlateau(monitor="val_loss",factor=0.5,patience=3,min_delta=0.001,min_lr=1e-5,mode="min",verbose=1)
# Save weights from the epoch with lowest validation loss
checkpoint = tf.keras.callbacks.ModelCheckpoint(filepath="results/model_a_adaptive_lr.weights.h5",monitor="val_loss",save_best_only=True,save_weights_only=True,mode="min",verbose=1)
history_adaptive_adam_001 = ModelA.fit(X_train,y_train,validation_data=(X_val, y_val),epochs=20,batch_size=64,shuffle=True,callbacks=[lr_scheduler, checkpoint],verbose=1)
```

```python
val_probs = ModelA.predict(X_val, batch_size=256, verbose=0)
val_pred = np.argmax(val_probs, axis=1)
val_true = np.asarray(y_val).reshape(-1)

print("Macro Precision:", precision_score(
    val_true, val_pred, average="macro", zero_division=0
))

print("Macro Recall:", recall_score(
    val_true, val_pred, average="macro", zero_division=0
))

print("Macro F1:", f1_score(
    val_true, val_pred, average="macro", zero_division=0
))

print("Macro Accuracy:", accuracy_score(
    val_true, val_pred
))

print("\nClassification Report:")
print(classification_report(
    val_true,
    val_pred,
    labels=[0, 1, 2, 3],
    target_names=["CNV", "DME", "Drusen", "Normal"],
    zero_division=0
))
```

```python
best_model = build_model(verbose=False)

best_model.load_weights(
    "results/model_a_adaptive_lr.weights.h5"
)

best_probs = best_model.predict(
    X_val,
    batch_size=256,
    verbose=0
)

best_preds = np.argmax(best_probs, axis=1)
val_true = np.asarray(y_val).reshape(-1)

print("Best Checkpoint Macro-F1:",
      f1_score(
          val_true,
          best_preds,
          average="macro",
          zero_division=0
      ))

print(classification_report(
    val_true,
    best_preds,
    labels=[0, 1, 2, 3],
    target_names=["CNV", "DME", "Drusen", "Normal"],
    zero_division=0
))
```

*Final Test Set Evaluation:*

```python
# Prepare directories for evaluation results
Path("results/metrics").mkdir(parents=True, exist_ok=True)
Path("results/figures").mkdir(parents=True, exist_ok=True)

# Prepare test labels
y_test_final = np.asarray(y_test).reshape(-1)

# Class names
class_names = ["CNV", "DME", "Drusen", "Normal"]

# Use the final-epoch trained models
models = {
    "Model A": ModelA,
    "Model B": ModelB
}

print("Test dataset:", X_test.shape)
print("Test labels:", y_test_final.shape)

for name, model in models.items():
    print(f"{name}: {model.count_params():,} parameters")
```

```python
from sklearn.metrics import (accuracy_score,precision_score,recall_score,f1_score,classification_report,confusion_matrix,ConfusionMatrixDisplay)

test_results = []
confusion_matrices = {}

for name, model in models.items():

    print(f"\n{'=' * 55}")
    print(f"FINAL TEST EVALUATION — {name}")
    print(f"{'=' * 55}")

    # Predict class probabilities
    y_prob = model.predict(
        X_test,
        batch_size=128,
        verbose=1
    )

    # Convert probabilities to predicted class labels
    y_pred = np.argmax(y_prob, axis=1)

    # Calculate metrics
    accuracy = accuracy_score(y_test_final, y_pred)

    precision = precision_score(
        y_test_final,
        y_pred,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        y_test_final,
        y_pred,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        y_test_final,
        y_pred,
        average="macro",
        zero_division=0
    )

    # Generate detailed classification report
    report = classification_report(
        y_test_final,
        y_pred,
        labels=[0, 1, 2, 3],
        target_names=class_names,
        zero_division=0
    )

    print("\nClassification Report:")
    print(report)

    # Save classification report as CSV
    report_dict = classification_report(
        y_test_final,
        y_pred,
        labels=[0, 1, 2, 3],
        target_names=class_names,
        output_dict=True,
        zero_division=0
    )

    filename = name.lower().replace(" ", "_")

    pd.DataFrame(report_dict).transpose().to_csv(
        f"results/metrics/{filename}_classification_report.csv"
    )

    # Calculate confusion matrix
    cm = confusion_matrix(
        y_test_final,
        y_pred,
        labels=[0, 1, 2, 3]
    )

    confusion_matrices[name] = cm

    # Store test metrics
    test_results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Macro Precision": precision,
        "Macro Recall": recall,
        "Macro F1": f1
    })

# Create comparison DataFrame
test_results_df = pd.DataFrame(test_results)

# Save results
test_results_df.to_csv(
    "results/metrics/final_test_results.csv",
    index=False
)

display(test_results_df.round(4))
```

```python
for name, cm in confusion_matrices.items():

    filename = name.lower().replace(" ", "_")

    # Save raw confusion matrix
    pd.DataFrame(
        cm,
        index=class_names,
        columns=class_names
    ).to_csv(
        f"results/metrics/{filename}_confusion_matrix.csv"
    )

    # Calculate row-normalized confusion matrix
    cm_normalized = cm.astype(float) / np.maximum(
        cm.sum(axis=1, keepdims=True), 1
    )

    fig, axes = plt.subplots(1, 2, figsize=(13, 5))

    # Raw confusion matrix
    disp1 = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=class_names
    )

    disp1.plot(
        ax=axes[0],
        cmap="Blues",
        values_format="d",
        colorbar=False
    )

    axes[0].set_title(f"{name} - Confusion Matrix")

    # Normalized confusion matrix
    disp2 = ConfusionMatrixDisplay(
        confusion_matrix=cm_normalized,
        display_labels=class_names
    )

    disp2.plot(
        ax=axes[1],
        cmap="Blues",
        values_format=".2f",
        colorbar=False
    )

    axes[1].set_title(
        f"{name} - Normalized Confusion Matrix"
    )

    plt.tight_layout()

    plt.savefig(
        f"results/figures/{filename}_confusion_matrices.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close(fig)
```
