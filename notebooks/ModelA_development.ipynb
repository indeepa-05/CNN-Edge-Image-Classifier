```python
import tensorflow as tf
import tensorflow.keras.layers as tfl

model_a = tf.keras.Sequential(name="Model_A_Standard_CNN")
```

```python
model_a.add(tf.keras.Input(shape=(64, 64, 1)))
model_a.add(tfl.Rescaling(1./255))
# Block 1: Standard convolution + max pooling
model_a.add(tfl.Conv2D(filters=32,kernel_size=(3, 3),strides=(1, 1),padding="same",activation="relu",name='Convolution1'))
model_a.add(tfl.MaxPooling2D(pool_size=(2, 2),strides=(2, 2),name='MaxPooling1'))
# Block 2: Standard convolution + max pooling
model_a.add(tfl.Conv2D(filters=64,kernel_size=(3, 3),strides=(1, 1),padding="same",activation="relu",name='Convolution2'))
model_a.add(tfl.MaxPooling2D(pool_size=(2, 2),strides=(2, 2),name='MaxPooling2'))
# Block 3: Standard convolution + max pooling
model_a.add(tfl.Conv2D(filters=128,kernel_size=(3, 3),strides=(1, 1),padding="same",activation="relu",name='Convolution3'))
model_a.add(tfl.MaxPooling2D(pool_size=(2, 2),strides=(2, 2),name='MaxPooling3'))
# Block 4: Standard convolution + max pooling
model_a.add(tfl.Conv2D(filters=256,kernel_size=(3, 3),strides=(1, 1),padding="same",activation="relu",name='Convolution4'))
model_a.add(tfl.MaxPooling2D(pool_size=(2, 2),strides=(2, 2),name='MaxPooling4'))
#Block 5: Global Average Pooling
model_a.add(tf.keras.layers.GlobalAveragePooling2D(name="GlobalAveragePooling"))
#Block 6: Dense Layer
model_a.add(tf.keras.layers.Dense(units=32,activation="relu",name="DenseHidden"))
#Block 7: Dense Layer
model_a.add(tf.keras.layers.Dense(units=4,activation="softmax",name="Output"))
```

```python
sample_images = X_train[:2]

predictions = model_a(sample_images, training=False)

print("Input shape:", sample_images.shape)
print("Output shape:", predictions.shape)
print("Predictions:\n", predictions.numpy())
```
